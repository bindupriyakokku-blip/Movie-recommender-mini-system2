```python
movies["features"] = (
    movies["genre"] + " " +
    movies["description"]
)

movies[["title", "language", "genre", "features"]]
```

```python
vectorizer = TfidfVectorizer(
    stop_words="english",
    ngram_range=(1, 2)
)

feature_matrix = vectorizer.fit_transform(movies["features"])

print("Feature Matrix Shape:", feature_matrix.shape)
```

```python
similarity_matrix = cosine_similarity(feature_matrix)

print("Similarity Matrix Created Successfully!")
```

```python
def recommend_movies(movie_name, number_of_recommendations=5):

    movie_name = movie_name.strip().lower()

    # Exact title search
    matching_movies = movies[
        movies["title"].str.lower() == movie_name
    ]

    # If exact match is not found
    if matching_movies.empty:

        # Partial title search
        matching_movies = movies[
            movies["title"].str.lower().str.contains(movie_name, na=False)
        ]

    if matching_movies.empty:
        print("❌ Movie not found in our dataset.")
        print("\nAvailable movies are:")

        for title in movies["title"]:
            print("•", title)

        return

    movie_index = matching_movies.index[0]

    selected_movie = movies.iloc[movie_index]

    similarity_scores = list(
        enumerate(similarity_matrix[movie_index])
    )

    # Sort from highest similarity to lowest
    similarity_scores = sorted(
        similarity_scores,
        key=lambda x: x[1],
        reverse=True
    )

    # Remove selected movie
    similarity_scores = [
        item for item in similarity_scores
        if item[0] != movie_index
    ]

    recommended_movies = similarity_scores[
        :number_of_recommendations
    ]

    print("\n" + "=" * 60)
    print("🎬 MOVIE RECOMMENDATION SYSTEM")
    print("=" * 60)

    print(f"\nSelected Movie : {selected_movie['title']}")
    print(f"Language       : {selected_movie['language']}")
    print(f"Genre          : {selected_movie['genre']}")

    print("\n⭐ Recommended Movies:\n")

    for rank, (index, score) in enumerate(
        recommended_movies, start=1
    ):

        movie = movies.iloc[index]

        print(
            f"{rank}. {movie['title']} "
            f"({movie['language']})"
        )

        print(f"   Genre      : {movie['genre']}")
        print(f"   Similarity : {score * 100:.2f}%")
        print()
```
